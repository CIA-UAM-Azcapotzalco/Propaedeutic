###3.9 PILAS Y COLAS

Nos enfocaremos en las dos estructuras de datos lineales más fundamentales: la Pila (Stack) y la Cola (Queue). Estas estructuras son esenciales porque controlan el orden de flujo de los datos, lo cual es la base para algoritmos más complejos como DFS y BFS

La Pila (Stack)

Definición Formal: Es una estructura de datos basada en el principio LIFO (Last-In, First-Out), lo que significa que el último elemento en entrar es el primero en salir.

Analogía para la clase: Imagina una pila de platos en una cafetería. Solo puedes poner un plato nuevo arriba de todos y, para limpiar uno, solo puedes agarrar el que está hasta arriba.

Formalidad Matemática y Complejidad:
Insertar (Push): O(1) - Solo se añade al final.

Eliminar (Pop): O(1) - Solo se retira el elemento superior.

Buscar: O(n) - En el peor caso, hay que vaciar toda la pila para encontrar un dato.



Ejemplo en Python para VS Code (Usando Listas):
En Python, las listas están optimizadas para actuar como pilas cuando usamos append() y pop().

````python
pila = []
# PUSH: Añadir elementos (van a la parte "superior")
pila.append("Plato 1")
pila.append("Plato 2")
pila.append("Plato 3")

print(f"Estado de la pila: {pila}") # Salida: ['Plato 1', 'Plato 2', 'Plato 3']

# POP: Quitar el último que entró
superior = pila.pop()
print(f"Se retiró: {superior}") # Salida: Plato 3
print(f"Pila actual: {pila}")    # Salida: ['Plato 1', 'Plato 2']
pila
````

La Cola (Queue)

Definición Formal: Es una estructura de datos basada en el principio FIFO (First-In, First-Out), donde el primer elemento en entrar es el primero en ser procesado.

Analogía para la clase: Es exactamente como una fila en el supermercado o en el banco. El cliente que llegó primero es el primero en ser atendido por el cajero.

Formalidad Matemática y Complejidad:

Insertar (Enqueue/Push): O(1) - Se añade al final de la fila.

Eliminar (Dequeue/Pop): O(1) - Se retira del frente.

Importante: En Python, usar lista.pop(0) para una cola es ineficiente (O(n)) porque hay que desplazar todos los demás elementos. Lo correcto es usar collections.deque.


Ejemplo en Python para VS Code (Usando deque):

````python
from collections import deque

# Inicializamos la cola
cola = deque(["Cliente A", "Cliente B"])

# ENQUEUE: Llega un nuevo cliente al final
cola.append("Cliente C")

# DEQUEUE: Atendemos al que está al frente (el primero que llegó)
atendido = cola.popleft() # ¡Ojo! popleft es O(1)

print(f"Atendiendo a: {atendido}") # Salida: Cliente A
print(f"Siguen en fila: {list(cola)}") # Salida: ['Cliente B', 'Cliente C']
cola
````

````python
def convertidor(cadena):
  conversion = []

  # Iterar sobre la lista (cadena) NO OLVIDES LA IDENTACIÓN (USA TAB PARA IDENTAR, NO ESPACIO)
  for dato in cadena:

  # Validar el tipo de dato para convertir
    if isinstance(dato, int):
    # Validar SI está dentro del rango, avisar al usuario sino (si el dato es entero)
      if dato >= 0 and dato <= 127:
        valor = entero_a_ascii(dato)
        conversion.append(valor)

    elif isinstance(dato, str):
      valor = ascii_a_entero(dato)
      conversion.append(valor)

  return conversion

````
